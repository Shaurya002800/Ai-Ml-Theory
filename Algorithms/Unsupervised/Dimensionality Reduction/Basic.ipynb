*Why Dimensionality Reduction Exists*

Real-world datasets are often very high dimensional.

Examples:

A 256×256 image → 65,536 features,
Genomic data → 20,000+ features,
Text (bag-of-words) → 50,000+ features

High dimensional data causes three major problems:

->Curse of Dimensionality

In high dimensions, points become almost equally distant from each other.
Distance-based algorithms like KNN, K-Means, and SVM stop working effectively because “near” and “far” lose meaning.

->Computational Cost

More features → more memory and longer training time.

Models with 10,000 features are much slower than models with 50 features.

->Overfitting

When features are too many compared to samples, models memorize noise instead of learning real patterns.

Key Insight:
Real data usually lies on a lower-dimensional structure (manifold) inside the high-dimensional space.

Example:
Face images with 65,536 pixels actually vary along only ~50–100 meaningful factors (lighting, pose, expression, identity).

Dimensionality Reduction finds these important factors and removes unnecessary dimensions.
