Key FastAPI Concepts

| Concept  | Purpose                               |
| -------- | ------------------------------------- |
| FastAPI  | Web framework for building APIs       |
| Endpoint | URL that performs an action           |
| GET      | Retrieve information                  |
| POST     | Send data for processing              |
| Pydantic | Validates incoming data automatically |
| JSON     | Standard format for API communication |
| Uvicorn  | Runs the FastAPI server               |


Important Interview Question
Why not Flask?

| Flask              | FastAPI              |
| ------------------ | -------------------- |
| More boilerplate   | Less boilerplate     |
| Manual validation  | Automatic validation |
| No built-in docs   | Automatic `/docs`    |
| Slower development | Faster development   |
| Common in web apps | Standard for ML APIs |


The Three Endpoints You'll Build
1. Health Check
GET /health

Purpose:

Verify server is running
Used by deployment platforms for monitoring
No ML inference involved

Example response:

{
  "status": "ok"
}
2. Single Prediction
POST /predict

Input:

{
  "text": "This movie was fantastic."
}

Output:

{
  "sentiment": "positive",
  "confidence": 0.97
}

Most common endpoint in production.

3. Batch Prediction
POST /predict/batch

Input:

{
  "texts": [
    "Amazing movie",
    "Terrible acting"
  ]
}

Output:

[
  {
    "sentiment": "positive",
    "confidence": 0.96
  },
  {
    "sentiment": "negative",
    "confidence": 0.99
  }
]

Useful when processing thousands of texts efficiently.

This app.py file turns your fine-tuned DistilBERT model into a production-ready REST API by loading the model once at startup, validating requests with Pydantic, exposing prediction endpoints, and returning structured JSON responses with confidence scores and latency metrics.


```python
# In your terminal (not notebook):
uvicorn app:app --reload --host 0.0.0.0 --port 8000

# You should see:
# INFO: Uvicorn running on http://0.0.0.0:8000
# INFO: Loading model...
# INFO: Model loaded.
```

```python
# Health check
curl http://localhost:8000/health

# Single prediction
curl -X POST http://localhost:8000/predict \
     -H "Content-Type: application/json" \
     -d '{"text": "This is genuinely one of the best films ever made."}'

# Batch prediction
curl -X POST http://localhost:8000/predict/batch \
     -H "Content-Type: application/json" \
     -d '{"texts": ["Loved every minute.", "Absolute rubbish.", "It was fine."]}'
```

```python
import requests

BASE = "http://localhost:8000"

# Single
r = requests.post(f"{BASE}/predict",
                  json={"text": "A stunning visual achievement."})
print(r.json())

# Batch
r = requests.post(f"{BASE}/predict/batch",
                  json={"texts": [
                      "Loved it.", "Hated it.", "Meh."
                  ]})
for pred in r.json()["predictions"]:
    print(f"[{pred['sentiment']:8s} {pred['confidence']:.1%}]  {pred['text']}")
```
